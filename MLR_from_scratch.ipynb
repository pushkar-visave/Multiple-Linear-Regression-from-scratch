```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
class MLR:
    def __init__(self):
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        # Insert a column of 1s at index 0 of X_train to account for the intercept (c)[cite: 6]
        X_train_bias = np.insert(X_train, 0, 1, axis=1)
        
        # Calculate coefficients using the Normal Equation: Beta = (X^T * X)^-1 * X^T * Y[cite: 6]
        # This utilizes np.linalg.inv and dot products to find the exact algebraic solution[cite: 6]
        betas = np.linalg.inv(np.dot(X_train_bias.T, X_train_bias)).dot(X_train_bias.T).dot(y_train)
        
        self.intercept_ = betas[0]
        self.coef_ = betas[1:]

    def predict(self, X_test):
        # Predict using y = X * Beta + c[cite: 6]
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
custom_mlr = MLR()
custom_mlr.fit(X_train, y_train)
custom_preds = custom_mlr.predict(X_test)
```

```python
sk_mlr = LinearRegression()
sk_mlr.fit(X_train, y_train)
sk_preds = sk_mlr.predict(X_test)
```

```python
print("Parameter Comparison-")
print(f"Custom MLR Intercept:   {custom_mlr.intercept_:.8f}")
print(f"Scikit-Learn Intercept: {sk_mlr.intercept_:.8f}")
```

```python
print("Coefficients Comparison (First 5)-")
print(f"Custom MLR Coefs:   {np.round(custom_mlr.coef_[:5], 4)}")
print(f"Scikit-Learn Coefs: {np.round(sk_mlr.coef_[:5], 4)}")
```

```python
print("Performance Comparison (R-Squared)-")
print(f"Custom MLR R2:   {r2_score(y_test, custom_preds):.8f}")
print(f"Scikit-Learn R2: {r2_score(y_test, sk_preds):.8f}")
```
